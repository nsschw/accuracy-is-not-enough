```python
import json
```

```python
with open("../data/processed/next-like-prediction-test-filtered.json", "r") as f:
    data = json.load(f)
```

```python
def create_prompts(data, max_num_samples=10, system_prompt="You are a social media user evaluating online posts. Keep your binary predictions (1 for like, 0 for dislike) consistent with your previous interaction history and the preferences you have expressed earlier."):
    prompts = []
    
    for user in data:
        prompt = []
        prompt.append({"role": "system", "content": system_prompt})
        
        for sample in user["input"]:
            prompt.append({"role": "user", "content": sample["text"]})
            prompt.append({"role": "assistant", "content": str(sample["label"])})

        for i, sample in enumerate(user["output"]):
            if i >= max_num_samples:
                break
            prompt.append({"role": "user", "content": sample["text"]})
            prompt.append({"role": "assistant", "content": str(sample["label"])})
            
        prompts.append({"messages": prompt.copy()})

    return prompts
```

```python
prompts = create_prompts(data)

with open("../data/processed/nlp-filtered-10-decoder-test-data.json", "w") as f:
    json.dump(prompts, f, indent=4)

prompts = create_prompts(data, max_num_samples=20)
with open("../data/processed/nlp-filtered-20-decoder-test-data.json", "w") as f:
    json.dump(prompts, f, indent=4)

prompts = create_prompts(data, max_num_samples=40)
with open("../data/processed/nlp-filtered-40-decoder-test-data.json", "w") as f:
    json.dump(prompts, f, indent=4)

prompts = create_prompts(data, max_num_samples=60)
with open("../data/processed/nlp-filtered-60-decoder-test-data.json", "w") as f:
    json.dump(prompts, f, indent=4)

prompts = create_prompts(data, max_num_samples=70)
with open("../data/processed/nlp-filtered-70-decoder-test-data.json", "w") as f:
    json.dump(prompts, f, indent=4)
```
